## Data Type Correction

### Why?

The `TotalCharges` column represents the total amount paid by a customer during their subscription period.

However, it is stored as an **object** instead of a numerical data type. This usually happens because the dataset contains blank spaces or non-numeric values.

Machine Learning algorithms cannot perform mathematical operations on object data types. Therefore, we first convert this column into a numeric format before proceeding with further analysis.

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

**Observation**

**Reason**

**Action**

Since these records are valid and contain useful information, replace the missing `TotalCharges` values with `0`.

## Feature 1 : Total Services

**Reason**

Customers using more services are generally more engaged with the company and may have different churn behavior.

**Action**

Count the total number of subscribed services for each customer.

```python
service_cols = [
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df["TotalServices"] = df[service_cols].apply(
    lambda x: sum(x.isin(["Yes","DSL","Fiber optic"])),
    axis=1
)

df[["TotalServices"]].head()
```

## Feature 2 : Average Monthly Spend

**Reason**

Average monthly spending helps identify customers with higher financial value.

**Action**

Calculate the average monthly spending using Total Charges and Tenure.

```python
df["AvgMonthlySpend"] = np.where(
    df["tenure"]==0,
    0,
    df["TotalCharges"]/df["tenure"]
)

df[["AvgMonthlySpend"]].head()
```

## Feature 3 : Customer Lifetime Value

**Reason**

Customers with higher lifetime value contribute more revenue and are more valuable to retain.

**Action**

Estimate customer lifetime value using monthly charges and tenure.

```python
df["CustomerLifetimeValue"] = (
    df["MonthlyCharges"] * df["tenure"]
)

df[["CustomerLifetimeValue"]].head()
```

## Feature 4 : Long-Term Customer

**Reason**

Customers who stay with the company for a longer period generally have lower churn risk.

**Action**

Classify customers with tenure greater than or equal to 24 months as long-term customers.

```python
df["LongTermCustomer"] = np.where(
    df["tenure"]>=24,
    1,
    0
)

df[["LongTermCustomer"]].head()
```

## Feature 5 : High-Value Customer

**Reason**

Customers paying higher monthly charges contribute more revenue.

**Action**

Identify customers whose monthly charges are above the 75th percentile.

```python
threshold = df["MonthlyCharges"].quantile(0.75)

df["HighValueCustomer"] = np.where(
    df["MonthlyCharges"]>=threshold,
    1,
    0
)

df[["HighValueCustomer"]].head()
```

```python
X = df.drop("Churn", axis=1)
y = df["Churn"]
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoders = {}

X = df.drop("Churn", axis=1)
y = df["Churn"]

for col in X.select_dtypes(include="object").columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
    label_encoders[col] = dict(zip(le.classes_, le.transform(le.classes_)))

target_encoder = LabelEncoder()
y = target_encoder.fit_transform(y)
```

```python
X.head()
```

```python
X.info()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Shape :", X_train.shape)
print("Testing Shape :", X_test.shape)
```

# Logistic Regression

Train the baseline classification model and evaluate its performance.

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=10000, random_state=42)

lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)

print("Accuracy :", accuracy_score(y_test, y_pred_lr))
print("Precision :", precision_score(y_test, y_pred_lr))
print("Recall :", recall_score(y_test, y_pred_lr))
print("F1 Score :", f1_score(y_test, y_pred_lr))
print("ROC-AUC :", roc_auc_score(y_test, y_pred_lr))
print("\n\n\nClassification Report:\n")
print(classification_report(y_test, y_pred_lr))
# Confusion Matrix
plt.figure(figsize=(6,4))

sns.heatmap(
    confusion_matrix(y_test, y_pred_lr),
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# Decision Tree
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)

print("Accuracy :", accuracy_score(y_test, y_pred_dt))
print("Precision :", precision_score(y_test, y_pred_dt))
print("Recall :", recall_score(y_test, y_pred_dt))
print("F1 Score :", f1_score(y_test, y_pred_dt))
print("ROC-AUC :", roc_auc_score(y_test, y_pred_dt))

print("\n\n\nClassification Report:\n")
print(classification_report(y_test, y_pred_dt))
plt.figure(figsize=(6,4))

sns.heatmap(
    confusion_matrix(y_test, y_pred_dt),
    annot=True,
    fmt="d",
    cmap="Oranges"
)

plt.title("Decision Tree - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
#  Random Forest

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

print("Accuracy :", accuracy_score(y_test, y_pred_rf))
print("Precision :", precision_score(y_test, y_pred_rf))
print("Recall :", recall_score(y_test, y_pred_rf))
print("F1 Score :", f1_score(y_test, y_pred_rf))
print("ROC-AUC :", roc_auc_score(y_test, y_pred_rf))
print("\n\n\nClassification Report:\n")
print(classification_report(y_test, y_pred_rf))
plt.figure(figsize=(6,4))

sns.heatmap(
    confusion_matrix(y_test, y_pred_rf),
    annot=True,
    fmt="d",
    cmap="Greens"
)

plt.title("Random Forest - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

# Model Comparison

The performance of all machine learning models is compared using multiple evaluation metrics to identify the most suitable model for customer churn prediction.

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy_score(y_test,y_pred_lr),
        accuracy_score(y_test,y_pred_dt),
        accuracy_score(y_test,y_pred_rf)
    ],
    "Precision": [
        precision_score(y_test,y_pred_lr),
        precision_score(y_test,y_pred_dt),
        precision_score(y_test,y_pred_rf)
    ],
    "Recall": [
        recall_score(y_test,y_pred_lr),
        recall_score(y_test,y_pred_dt),
        recall_score(y_test,y_pred_rf)
    ],
    "F1 Score": [
        f1_score(y_test,y_pred_lr),
        f1_score(y_test,y_pred_dt),
        f1_score(y_test,y_pred_rf)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test,y_pred_lr),
        roc_auc_score(y_test,y_pred_dt),
        roc_auc_score(y_test,y_pred_rf)
    ]
})

comparison = comparison.round(3)

comparison
```

### Conclusion

Among the three models, **Logistic Regression** achieved the highest Accuracy, F1-Score, and ROC-AUC score. Therefore, it is selected as the final model for customer churn prediction.

# Hyperparameter Tuning

GridSearchCV was used to identify the optimal hyperparameters for the Logistic Regression model and improve its performance.

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "C":[0.01,0.1,1,10],
    "solver":["liblinear","lbfgs"]
}

grid = GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid,
    cv=5,
    scoring="f1"
)

grid.fit(X_train,y_train)

print("Best Parameters :",grid.best_params_)
print("Best Score :",grid.best_score_)
```

```python
best_lr = grid.best_estimator_

y_pred_best = best_lr.predict(X_test)
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred_best))
print("Precision :", precision_score(y_test, y_pred_best))
print("Recall :", recall_score(y_test, y_pred_best))
print("F1 Score :", f1_score(y_test, y_pred_best))
print("ROC-AUC :", roc_auc_score(y_test, y_pred_best))
```

```python
print(classification_report(y_test, y_pred_best))
```

```python
plt.figure(figsize=(6,4))

sns.heatmap(
    confusion_matrix(y_test, y_pred_best),
    annot=True,
    fmt="d",
    cmap="Greens"
)

plt.title("Tuned Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
```

<h1 align="center">Customer Churn Prediction</h1>

The trained Logistic Regression model is used to predict whether a randomly selected customer is likely to churn. Along with the prediction, the model also provides the probability of staying and churning using `predict_proba()`.

```python
encoding_table = []

for col, mapping in label_encoders.items():
    for category, value in mapping.items():
        encoding_table.append([col, category, value])

encoding_df = pd.DataFrame(
    encoding_table,
    columns=["Column", "Category", "Encoded Value"]
)
print("For Understanding Encoded Values: ")
encoding_df
```

```python
# Select a random customer
sample = X.sample(1, random_state=None)

print("Customer Details Used For Prediction")
print(sample.T)

# Prediction
prediction = best_lr.predict(sample)[0]

# Prediction Probability
prob = best_lr.predict_proba(sample)[0]

# Display Result
if prediction == 1:
    print("\nPrediction : Customer Will Churn")
else:
    print("\nPrediction : Customer Will Stay")

print(f"Stay Probability  : {prob[0]*100:.2f}%")
print(f"Churn Probability : {prob[1]*100:.2f}%")
```

```python
#  Manual Values for prediction

new_customer = pd.DataFrame({
    "gender":[1],
    "SeniorCitizen":[0],
    "Partner":[1],
    "Dependents":[0],
    "tenure":[5],
    "PhoneService":[1],
    "MultipleLines":[0],
    "InternetService":[1],
    "OnlineSecurity":[0],
    "OnlineBackup":[1],
    "DeviceProtection":[0],
    "TechSupport":[0],
    "StreamingTV":[1],
    "StreamingMovies":[1],
    "Contract":[0],
    "PaperlessBilling":[1],
    "PaymentMethod":[2],
    "MonthlyCharges":[95.6],
    "TotalCharges":[478],
    "TotalServices":[6],
    "AvgMonthlySpend":[95.6],
    "CustomerLifetimeValue":[478],
    "LongTermCustomer":[0],
    "HighValueCustomer":[1]
})
```

```python
prediction = best_lr.predict(new_customer)

probability = best_lr.predict_proba(new_customer)

if prediction[0] == 1:
    print("⚠️ Customer is likely to Churn")
else:
    print("✅ Customer is likely to Stay")

print(f"Churn Probability : {probability[0][1]*100:.2f}%")
```

# ROC Curve

The ROC Curve evaluates the model's ability to distinguish between churn and non-churn customers. A curve closer to the top-left corner indicates better model performance.

```python
from sklearn.metrics import roc_curve, auc

y_prob = best_lr.predict_proba(X_test)[:,1]

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8,6))

plt.plot(fpr, tpr,
         color="#2E86DE",
         linewidth=2,
         label=f"AUC = {roc_auc:.3f}")

plt.plot([0,1],[0,1],'k--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()

plt.grid(alpha=0.3)
plt.show()
```

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": best_lr.coef_[0]
})

feature_importance["Absolute"] = feature_importance["Coefficient"].abs()

feature_importance = feature_importance.sort_values(
    by="Absolute",
    ascending=False
)

feature_importance.head(15)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=feature_importance.head(10),
    x="Coefficient",
    y="Feature",
    palette="viridis"
)

plt.title("Top 10 Important Features", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```
